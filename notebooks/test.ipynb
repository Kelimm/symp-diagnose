```python
import pandas as pd
data_path = "../data/Diseases_and_Symptoms_dataset.csv"
data = pd.read_csv(data_path)
data
```

```python
y = data.iloc[:, 0] 
X = data.drop(columns=["diseases"])
print(X.shape)
y.shape
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

y_encoded = le.fit_transform(y)
len(le.classes_)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics 



X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size = 0.2, random_state=42, stratify=y_encoded)


model = LogisticRegression(random_state=42)


model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(f"Précision = {metrics.accuracy_score(y_test, y_pred) * 100:.2f}%")
```

```python
from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier(200, min_samples_leaf=5)
clf.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(f"Précision = {metrics.accuracy_score(y_test, y_pred) * 100:.2f}%")
```

```python
import joblib
# joblib.dump(model, '../models/logistic_regression.pkl')
joblib.dump(clf, '../models/Random_Forest.pkl')
```

```python
joblib.dump(le, '../models/label_encoder.pkl')

```

```python
from xgboost import XGBClassifier

clf = XGBClassifier(max_depth=5, learning_rate = 0.2, n_estimators = 150, n_jobs=-1)
clf = clf.fit(X_train, y_train)

predict = clf.predict(X_test)
print(f"Précision = {metrics.accuracy_score(y_test, predict) * 100:.2f}%")
```

```python
from sklearn.metrics import classification_report

rapport = classification_report(
    y_test, 
    y_pred, 
    target_names=le.classes_, 
    zero_division=0 
)
print(rapport)
```
