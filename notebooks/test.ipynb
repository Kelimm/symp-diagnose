```python
import pandas as pd
data_path = "../data/Diseases_and_Symptoms_dataset.csv"
data = pd.read_csv(data_path)
data
```

```python
y = data.iloc[:, 0] 
X = data.drop(columns=["diseases"])
print(X.shape)
y.shape
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

y_encoded = le.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn import metrics 
from sklearn.ensemble import VotingClassifier

X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size = 0.2, random_state=42, stratify=y_encoded)

modele = LogisticRegression(random_state=42)

modele = modele.fit(X_train, y_train)
y_pred = modele.predict(X_test)

print(f"Précision = {metrics.accuracy_score(y_test, y_pred) * 100:.2f}%")
```

```python
import joblib
joblib.dump(modele, '../models/logistic_regression.pkl')
```

```python
joblib.dump(le, '../models/label_encoder.pkl')
```
